# Analise de Dataset

A empresa Umbrella Corporation contratou desenvolvedores Python (🐍) para analisar um dataset para verificar o alto número de cancelamentos de contratos dessa empresa, e para entregar uma solução para este problema.

Para isso, a Umbrella Corp. forneceu um **dataset** com os detalhes dos cancelamentos. Os Pythonianos (🐍)  devem analisar o arquivo utilizando o **Pandas** (💀) .

Instalar o **Pandas** (💀):
- pip install pandas

Vamos começar trazendo todos os dados do **dataset**.


```python
import pandas as pd
tabela = pd.read_csv("cancelamentos.csv")
display(tabela)
```

## Tratamento de Dados

Ao ler o **dataset**, precebemos que existem células com valores nulos, que atrapalham a nossa análise, e portanto devem ser retirados da análise.

Vamos eliminá-los e exibir os valores novamente:

```python
tabela = tabela.drop("CustomerID", axis=1)
display(tabela)
```

Vamos pedir para que o **Python** (🐍) nos retorne o número de linhas preenchidas da tabela.

```python
display(tabela.info())
```

Veja que temos em cada uma das colunas, um número diferente de linhas. Isso acontece porque temos dados nulos na tabela, e que devemos eliminar.

Portanto, vamos retirar através do comando **Phyton** (🐍) abaixo:

```python
tabela = tabela.dropna()
display(tabela.info())
```

# Analise de Dados

Agora sim é que começa a análise de dados propriamente dita.

Vamos começar verificando o número de Clientes que cancelaram o contrato e o número de clientes que não cancelaram.

```python
display(tabela["cancelou"].value_counts())
```

Podemos perceber que o número de clientes que cancelaram realmente é alto, inclusive mais alto que o número de clientes que não cancelaram, mas quanto que isso da em porcentagem?

Vamos descobrir:

```python
display(tabela["cancelou"].value_counts(normalize=True).map("{:.2%}".format))
```
