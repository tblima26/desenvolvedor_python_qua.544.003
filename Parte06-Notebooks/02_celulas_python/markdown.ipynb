Hoje vamos explorar o uso de Células Python em nossos notebooks.

As células de código são blocos interativos onde você pode escrever, testar e executar códigos Python diretamente no seu ambiente de trabalho, unindo texto explicativo e programação em um só lugar.

Observe o exemplo na célula abaixo:

```python
nome = "😎 Thiago Borges"
texto = f"Esta é uma célula Pyhton. Escrita pelo melhor de todos:  {nome} "

print(texto)
```

```python
n1 = int(input('Informe um número: '))
print(f'Valor da variavél é {n1}')
```

O valor faz sentido quando é informado diretamente. No entanto, analise a próxima célula e observe: o valor se mantém e se repete ao utilizarmos a mesma variável?

```python
resultado = n1
print(resultado)
```

Sim, o valor se mantém armazenado na sessão, mas lembre-se: se você reatribuir um novo valor, o anterior será substituído.

Além disso, fique atento à **ordem de execução**, não apenas à posição visual das células. 

Veja este exemplo:
* **Célula 1:** `x = 10` *(⚠️ NÃO EXECUTADA)*
* **Célula 2:** `result = x + 20` *(EXECUTADA)*
* **Célula 3:** `print(result)` *(EXECUTADA)*

Se rodar a 2 e a 3 sem executar a 1, o Python dará um erro (`NameError`), pois a variável `x` nunca foi criada na memória. **O que importa é o que foi rodado, e não apenas o que está escrito!**
